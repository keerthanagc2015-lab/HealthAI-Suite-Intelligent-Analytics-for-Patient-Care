```python
#Bivarate analysis
#Primary Diagnosis vs HbA1c

plt.figure(figsize=(14, 7))

median_hba1c = (
    df.groupby("Primary_Diagnosis")["HbA1c_%"]
      .median()
      .sort_values(ascending=False)
)

order = median_hba1c.index

ax = sns.boxplot(
    data=df,
    x="Primary_Diagnosis",
    y="HbA1c_%",
    order=order
)

# Add median values
for i, diagnosis in enumerate(order):
    median_value = median_hba1c[diagnosis]

    ax.text(
        i,
        median_value,
        f"{median_value:.1f}",
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.title("HbA1c Distribution by Primary Diagnosis")
plt.xlabel("Primary Diagnosis")
plt.ylabel("HbA1c (%)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
### Business Insight

#Type 2 Diabetes patients show a substantially higher median HbA1c (8.2%) compared with other diagnosis groups (around 5.6–5.8%). This indicates HbA1c is a strong differentiating clinical feature for diabetes-related patient identification and risk assessment.
```

```python
#Diagnosis vs Blood Glucose

plt.figure(figsize=(14, 7))

median_glucose = (
    df.groupby("Primary_Diagnosis")["Blood_Glucose_mg_dL"]
      .median()
      .sort_values(ascending=False)
)

order = median_glucose.index

ax = sns.boxplot(
    data=df,
    x="Primary_Diagnosis",
    y="Blood_Glucose_mg_dL",
    order=order
)

# Add median values to each box
for i, diagnosis in enumerate(order):

    median_value = median_glucose[diagnosis]

    ax.text(
        i,
        median_value,
        f"{median_value:.1f}",
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.title("Blood Glucose Distribution by Primary Diagnosis")
plt.xlabel("Primary Diagnosis")
plt.ylabel("Blood Glucose (mg/dL)")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
### Business Insight

#Type 2 Diabetes patients have a substantially higher median blood glucose level (179.8 mg/dL) compared with other diagnosis groups (approximately 104–110 mg/dL). This suggests blood glucose is a strong differentiating feature for identifying diabetes-related patients and can be valuable for risk prediction.
```

```python
#Treatment Type vs Treatment Outcome
plt.figure(figsize=(14, 7))

ax = sns.countplot(
    data=df,
    x="Treatment_Type",
    hue="Treatment_Outcome"
)

# Add values on bars
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%d",
        fontsize=8,
        padding=2
    )

plt.title("Treatment Outcomes by Treatment Type")
plt.xlabel("Treatment Type")
plt.ylabel("Number of Patients")

plt.xticks(rotation=30, ha="right")

plt.legend(
    title="Treatment Outcome",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
#Hospital Type vs Treatment Outcome (%)

# Create percentage table
hospital_outcome = pd.crosstab(
    df["Hospital_Type"],
    df["Treatment_Outcome"],
    normalize="index"
) * 100

hospital_outcome
```

```python
### Business Insight

#Treatment outcomes are almost evenly distributed at approximately 20% across Corporate, Government, and Private hospitals. This suggests that hospital type alone does not show a strong association with treatment outcomes in this dataset.
```

```python
#Convert Visit Date

df["Visit_Date"] = pd.to_datetime(df["Visit_Date"])

print(df["Visit_Date"].dtype)
print("Earliest Visit:", df["Visit_Date"].min())
print("Latest Visit:", df["Visit_Date"].max())
```

```python
#Create time features

df["Year"] = df["Visit_Date"].dt.year
df["Month"] = df["Visit_Date"].dt.month
df["Year_Month"] = df["Visit_Date"].dt.to_period("M")

print(df[["Visit_Date", "Year", "Month", "Year_Month"]].head())
```

```python
#Visit Month vs Clinical Values

monthly_clinical = (
    df.groupby("Year_Month")[
        [
            "BMI",
            "Blood_Glucose_mg_dL",
            "HbA1c_%",
            "Total_Cholesterol_mg_dL"
        ]
    ]
    .mean()
    .round(2)
)

monthly_clinical
```

```python
### Business Insight

#Average BMI, blood glucose, HbA1c, and cholesterol remain relatively stable across the study period, with only minor month-to-month fluctuations. No clear temporal trend in the overall clinical profile of patients is observed in this dataset.
```

```python
#Monthly Diagnosis Percentage Table

monthly_diagnosis = pd.crosstab(
    df["Year_Month"],
    df["Primary_Diagnosis"],
    normalize="index"
) * 100

monthly_diagnosis = monthly_diagnosis.round(2)

monthly_diagnosis
```

```python
### Business Insight

#The monthly diagnosis distribution remains relatively stable throughout the study period, with Type 2 Diabetes consistently being the most represented diagnosis followed by Hypertension. No clear long-term or seasonal shift in the overall diagnosis pattern is observed.
```
